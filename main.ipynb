## _Функція для обчислення функції втрат у векторному вигляді_


```python
import numpy as np

def hypothesis(X, w):
    return X @ w

def compute_cost(X, y, w):
    m = len(y)
    predictions = hypothesis(X, w)
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    
    return cost
```

## _Реалізація одного кроку градієнтного спуску_

```python
import numpy as np

def hypothesis(X, w):
    return X @ w


def gradient_descent_step(X, y, w, learning_rate):
    m = len(y)
    predictions = hypothesis(X, w)
    error = predictions - y
    gradient = (1 / m) * (X.T @ error)
    
    w_new = w - learning_rate * gradient
    
    return w_new
```

## _Знаходження найкращих параметрів w для датасету використовуючи написані вами функції, прогнозуючу ціну на будинок залежно від площі, кількості ванних кімнат та кількості спалень_

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

def compute_cost(X, y, w):
    m = len(y)
    predictions = hypothesis(X, w)
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return cost

def hypothesis(X, w):
    return X @ w

def gradient_descent_step(X, y, w, learning_rate):
    m = len(y)
    predictions = hypothesis(X, w)
    error = predictions - y
    gradient = (1 / m) * (X.T @ error)
    w_new = w - learning_rate * gradient
    return w_new

df = pd.read_csv('Housing.csv') 

X_raw = df[['area', 'bedrooms', 'bathrooms']].values
y = df['price'].values.reshape(-1, 1)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

m = len(y)
X_bias = np.hstack((np.ones((m, 1)), X_scaled))

np.random.seed(42) 
w_gradient_descent = np.random.randn(X_bias.shape[1], 1) * 0.01

learning_rate = 0.01   
iterations = 2000     

for i in range(iterations):
    w_gradient_descent = gradient_descent_step(X_bias, y, w_gradient_descent, learning_rate)

print(f"w0 (Вільний член / Intercept) : {w_gradient_descent[0][0]:.2f}")
print(f"w1 (Вага площі / Area)        : {w_gradient_descent[1][0]:.2f}")
print(f"w2 (Вага спалень / Bedrooms)  : {w_gradient_descent[2][0]:.2f}")
print(f"w3 (Вага ванних / Bathrooms)  : {w_gradient_descent[3][0]:.2f}")
print(f"Фінальна помилка (MSE)        : {compute_cost(X_bias, y, w_gradient_descent):.2f}")

```

## _Знаходження цих параметрів за допомогою аналітичного рішення_

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

def compute_cost(X, y, w):
    m = len(y)
    predictions = hypothesis(X, w)
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return cost

def hypothesis(X, w):
    return X @ w

def gradient_descent_step(X, y, w, learning_rate):
    m = len(y)
    predictions = hypothesis(X, w)
    error = predictions - y
    gradient = (1 / m) * (X.T @ error)
    w_new = w - learning_rate * gradient
    return w_new

df = pd.read_csv('Housing.csv') 

X_raw = df[['area', 'bedrooms', 'bathrooms']].values
y = df['price'].values.reshape(-1, 1)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

m = len(y)
X_bias = np.hstack((np.ones((m, 1)), X_scaled))

np.random.seed(42) 
w_gradient_descent = np.random.randn(X_bias.shape[1], 1) * 0.01

learning_rate = 0.01   
iterations = 2000     

for i in range(iterations):
    w_gradient_descent = gradient_descent_step(X_bias, y, w_gradient_descent, learning_rate)



w_analytical = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y

print("="*50)
print("ПАРАМЕТРИ W (ЗНАЙДЕНІ АНАЛІТИЧНИМ РІШЕННЯМ):")
print("="*50)
print(f"w0 (Intercept) : {w_analytical[0][0]:.2f}")
print(f"w1 (Area)      : {w_analytical[1][0]:.2f}")
print(f"w2 (Bedrooms)  : {w_analytical[2][0]:.2f}")
print(f"w3 (Bathrooms) : {w_analytical[3][0]:.2f}")
```

## _Перевірка через skilit-learn_

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(X_scaled, y)

w_sklearn = np.hstack((lr_model.intercept_.ravel(), lr_model.coef_.ravel()))

df_compare = pd.DataFrame({
    "Параметр": ["w0 (Intercept)", "w1 (Area)", "w2 (Bedrooms)", "w3 (Bathrooms)"],
    "Градієнтний спуск (NumPy)": w_gradient_descent.ravel(),
    "Аналітичне рішення": w_analytical.ravel(),
    "Scikit-Learn": w_sklearn
})

print("\n" + "="*60 + "\nФІНАЛЬНЕ ПОРІВНЯННЯ РЕЗУЛЬТАТІВ:\n" + "="*60)
print(df_compare.to_string(index=False))
```
